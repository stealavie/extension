```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('analytics.txt')
```

```python
df.head()
```

```python
sns.histplot(df['Latency'], bins=30, kde=True)  
```

```python
min_latency = df['Latency'].min()
max_latency = df['Latency'].max()
mean_latency = df['Latency'].mean()
std_latency = df['Latency'].std()
print(f"Min Latency: {min_latency} ms")
print(f"Max Latency: {max_latency} ms")
print(f"Mean Latency: {mean_latency:.2f} ms")
print(f"Std Dev Latency: {std_latency:.2f} ms")
```
